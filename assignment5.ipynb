```python
import requests

city = input("Enter city name: ")

url = f"https://api.openweathermap.org/data/2.5/weather?q={city}&appid=02922192b869edd75280ce67ed26b5b3&units=metric"

response = requests.get(url)

data = response.json()

if data["cod"] == 200:

    print("\n------ Weather Report ------")

    print("City:", data["name"])
    print("Country:", data["sys"]["country"])

    print("Temperature:", data["main"]["temp"], "°C")
    print("Feels Like:", data["main"]["feels_like"], "°C")

    print("Minimum Temp:", data["main"]["temp_min"], "°C")
    print("Maximum Temp:", data["main"]["temp_max"], "°C")

    print("Humidity:", data["main"]["humidity"], "%")
    print("Pressure:", data["main"]["pressure"])

    print("Weather:", data["weather"][0]["description"])

    print("Wind Speed:", data["wind"]["speed"], "m/s")

    print("Visibility:", data["visibility"])

    print("Sunrise:", data["sys"]["sunrise"])
    print("Sunset:", data["sys"]["sunset"])

else:
    print("Error:", data["message"])

    
```
