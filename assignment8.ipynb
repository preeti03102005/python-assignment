```python
# 2)Create two DataFrames, df1 and df2, with a common column (e.g., 'ID'). 

import pandas as pd
df1 = pd.DataFrame({
    'id': [1,2,3,4],
    'Name': ['Preeti', 'Sharad', 'Mohit', 'Aman']
})

df2 = pd.DataFrame({
    'id': [3,4,5,6],
    'Marks': [10,20,30,40]
})

print(df1)
print(df2)
```

```python
# >Perform an inner merge on this common column and display the resulting DataFrame.

inner_merge = pd.merge(df1, df2, on='id', how='inner')
print(inner_merge)
```

```python
# >Perform a left join of df1 and df2 on the 'ID' column. Explain how missing values are handled in the resulting DataFrame. Right Join and Index-Based Join.

left_join = pd.merge(df1, df2, on='id', how='left')
print(left_join)

right_join = pd.merge(df1, df2, on='id', how='right')
print(right_join)

df1_index = df1.set_index('id')
df2_index = df2.set_index('id')
joined_df = df1_index.join(df2_index)
print(joined_df)
```

```python
# >Perform a right join using pd.merge() on a common column, then perform a join using df.join() based on the index. Compare the results. Merging with Multiple Keys.

import pandas as pd
df3 = pd.DataFrame({
    'id': [1,2,3],
    'Name': ['Preeti', 'Sharad', 'Mohit'],
    'Salary': [10000,20000,30000]   
})

df4 = pd.DataFrame({
    'id': [3,2,4],
    'Name': ['Priya','Purva','Sakshi'],
    'Salary': [30000,20000,60000]
})

r_join = pd.merge(df3, df4, on='id', how='right')
print(r_join)

multi_merge = pd.merge(df3, df4, on=['id','Salary'], how='inner')
print(multi_merge)
```

```python
# 3) Create three DataFrames. Vertically concatenate two of them using pd.concat(), then merge the resulting DataFrame with the third DataFrame on a common key. T Understand join() vs. merge().

dfa = pd.DataFrame({
    'id': [1,2],
    'Name': ['Priya','Purva']
    
})

dfb = pd.DataFrame({
    'id': [3,4],
    'Name': ['Sakshi','Ravi']
   
})

dfc = pd.DataFrame({
    'id': [1,2,3,4],
    'Marks': [10,20,30,40]
})

concat_df = pd.concat([dfa, dfb], axis=0)
print(concat_df)
```

```python
final_merge = pd.merge(concat_df, dfc, on='id')
print(final_merge)
```

```python
# Also Explain the primary differences between df.join() and pd.merge(),

#1. merge()
# Used to combine DataFrames using columns.
# Supports:
# inner
# left
# right
# outer joins

#2. join()
# Mainly index-based joining.
# Easier syntax


#join
import pandas as pd

df1 = pd.DataFrame({
    'Name': ['Alice', 'Bob'],
    'Marks': [80, 90]
}, index=[1, 2])

df2 = pd.DataFrame({
    'City': ['Delhi', 'Mumbai']
}, index=[1, 2])

result = df1.join(df2)
print(result)


#merge
import pandas as pd

df1 = pd.DataFrame({
    'ID': [1, 2],
    'Name': ['Alice', 'Bob']
})

df2 = pd.DataFrame({
    'ID': [1, 2],
    'City': ['Delhi', 'Mumbai']
})

result = pd.merge(df1, df2, on='ID')
print(result)
```
