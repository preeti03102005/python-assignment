```python
#Assignment 5
#1) Create a CSV file for address book, CSV file should have column for name, address, mobile, email. Insert 2-3 dummy data entered by user.
#2)  Practice DATABASE  
#1) Create Database
#2) Create 2-3 tables
#3) Insert some records
#4) Perform different select operations
#5) Update some data
#6) Delete some data


import csv


data = []

n = int(input("How many records you want to enter: "))

for i in range(n):
    print("\nEnter Details for Person", i + 1)

    name = input("Enter Name: ")
    address = input("Enter Address: ")
    mobile = input("Enter Mobile Number: ")
    email = input("Enter Email: ")

    data.append([name, address, mobile, email])


with open("address_book.csv", "w", newline='') as file:

    writer = csv.writer(file)

    
    writer.writerow(["Name", "Address", "Mobile", "Email"])

    
    writer.writerows(data)

print("\nData inserted successfully into address_book.csv")
```

```python
import sqlite3

conn = sqlite3.connect("college.db")

conn.execute("""
CREATE TABLE student(
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    name TEXT,
    course TEXT,
    email TEXT
)
""")

print("Table Created Successfully")

conn.close()
```

```python
import sqlite3

conn = sqlite3.connect("college.db")

conn.execute("""
UPDATE student
SET course='Data Science'
WHERE name='Rahul'
""")

conn.commit()

print("Record Updated Successfully")

data = conn.execute("SELECT * FROM student")

for row in data:
    print(row)

conn.close()
```

```python
import sqlite3

conn = sqlite3.connect("college.db")

conn.execute("DELETE FROM student WHERE name='Aman'")

conn.commit()

print("Record Deleted Successfully")

data = conn.execute("SELECT * FROM student")

for row in data:
    print(row)

conn.close()
```
