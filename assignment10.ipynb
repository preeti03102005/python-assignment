```python
import numpy as np

a = np.array([
    [1,2,3,4],
    [np.nan,6,7,8]
])
print("\nOriginal array:\n",a)

a = np.nan_to_num(a,nan=0)
print("\nReplacing array:\n",a)

b=a.T
print("\nTranspose Array:\n",b)
```

```python
import numpy as np

a = np.arange(24).reshape(2,3,4)

print("\nOriginal Shape:\n",a.shape)
print(a)

b = np.moveaxis(a,0,2)
print("\nNew Shape:\n",b.shape)
```

```python
import numpy as np

a = np.array([
    [1,2,4],
    [np.nan,6,7],
    [7,8,np.nan]
])
print("\nOriginal array:\n",a)

col = np.nanmean(a, axis=0)
print("\ncolumn mean:\n", col)

inds = np.where(np.isnan(a))
a[inds] = np.take(col, inds[1])
print("\nArray:\n", a)
```
