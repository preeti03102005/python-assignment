```python
#> Create function for following

#1. Write a Python function to find the maximum of three numbers.

def max(a,b,c):
    if a >= b and a >= c:
        return a
    elif b >= a and b >= c:
        return b
    else:
        return c
    
print(max(10,20,30))
```

```python
#3. Write a Python function to multiply all the numbers in a list.


def multiply(numbers):
    result = 1
    for num in numbers:
        result = result * num
    return result
print(multiply([5,8,9]))
```

```python
#4. Write a Python function to calculate the factorial of a number (a non-negative integer). The function accepts the number as an argument.


def factorial(n):
    fact = 1
    for i in range(1, n+1):
        fact = fact * i
    return fact

print(factorial(7))
```

```python
#5. Write a Python program to reverse a string.

def reverse_string(text):
    return text[::-1]

print(reverse_string("Preeti"))
```

```python
#6. Write a Python function to check whether a number falls within a given range

def check_range(n):
    if n in range(1,11):
        print("no. is in range")
    else:
        print("no. is outside the range")
        
check_range(5)        
```

```python
#7. Write a Python function to Print Even Numbers from a Given List

def even_num(lst):
    for num in lst:
        if num%2 == 0:
            print(num)
            
even_num([1,2,3,4,5,6,7,8,9])            
```

```python
#8. Write a Python function that takes a number as a parameter and checks whether the number is prime or not.


def prime_num(n):
    if n>1:
        for i in range(2,n):
            if n%i == 0:
                print("not a prime no")
                break
        else:
            print("Prime Number")
    else:
        print("Not a Prime Number")

prime_num(7)
```

```python
#9. Write a Python function that accepts a string and counts the number of upper and lower case letters.

def count_letter(t):
    u = 0
    l = 0

    for ch in t:
        if ch.isupper():
            u += 1
        elif ch.islower():
            l += 1

    print("Upper case:", u)
    print("Lower case:", l)
    
count_letter("Hello World")
```
