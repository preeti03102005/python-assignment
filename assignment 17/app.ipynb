```python
from pyspark import SparkContext
```

```python
sc = SparkContext("local[*]", "EmployeeRDD")
```

```python
rdd = sc.textFile("data/employees.csv")
header = rdd.first()
```

```python
employees = (
    rdd.filter(lambda row: row != header)
       .map(lambda row: row.split(","))
       .map(lambda row: (
           int(row[0]),      # id
           row[1],           # name
           row[2],           # department
           int(row[3])       # salary
       ))
)
```

```python
sorted_emp = employees.sortBy(
    lambda emp: emp[3],
    ascending=False
)

print("Employees Sorted By Salary")

for emp in sorted_emp.collect():
    print(emp)

sorted_emp.map(
    lambda emp: f"{emp[0]},{emp[1]},{emp[2]},{emp[3]}"
).saveAsTextFile("output/sorted_employees")
```
