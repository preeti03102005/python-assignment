```python
import numpy as np
arr = np.array([
    [1,2,3,np.nan,5],
    [np.nan, -2, 5, 0,-7]
])

print(arr,"\n")

arr = np.nan_to_num(arr, nan=0)
print(arr,"\n")

arr_transpose = arr.transpose()
print(arr_transpose)
```

```python
import numpy as np
arr = np.array([
    [1,2,3,np.nan,5],
    [np.nan, -2, 5, 0,-7]
])

print(arr,"\n")

col_mean = np.nanmean(arr, axis=0)
inds = np.where(np.isnan(arr))
arr[inds] = np.take(col_mean, inds[1])
print(arr)
```

```python
import numpy as np
arr1 = np.array([[3, 4],[1,2]]) 
arr2 = np.array([[1, 0],[7,8]])
avg = (arr1 + arr2)/2
print("Average of NumPy arrays:\n", avg)


comb = np.concatenate((arr1.flatten(),
                           arr2.flatten())) 


print("Mean:", np.mean(comb))
print("Median:", np.median(comb))

values, counts = np.unique(comb, return_counts=True)
mode = values[np.argmax(counts)]

print("Mode:", mode)
```

```python
import numpy as np

a = np.array([
    [1,-2,3],
    [-1,3,-1],
    [2,-5,5]
])
b = np.array([9,-6,17])
X1 = np.linalg.solve(a,b)
print(X1)

a_inv = np.linalg.inv(a)

X2 = np.dot(a_inv,b)

print(X2)
```
