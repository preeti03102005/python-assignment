```python
import re

mobile = "9876543210"

pattern = r'^[6-9]\d{9}$'

if re.match(pattern, mobile):
    print("Valid Mobile Number")
else:
    print("Invalid Mobile Number")
```

```python
import pandas as pd

df = pd.DataFrame({
    'Date': pd.date_range('2026-01-01', periods=7),
    'Sales': [100, 150, 200, 250, 300, 350, 400]
})

print(df)

df['Day_Name'] = df['Date'].dt.day_name()

df['Month'] = df['Date'].dt.month

df['Rolling_Avg'] = df['Sales'].rolling(3).mean()

print("\nUpdated Data")
print(df)

print("\nAverage Sales")
print(df['Sales'].mean())
```

```python
import pandas as pd


df = pd.read_csv("tips.csv")

print("Original Dataset")
print(df)

print("\nShape of Dataset")
print(df.shape)

print("\nColumn Names")
print(df.columns)

print("\nData Types")
print(df.dtypes)

print("\nDataset Info")
print(df.info())

print("\nMissing Values")
print(df.isnull().sum())

print("\nDuplicate Rows")
print(df.duplicated().sum())

df.drop_duplicates(inplace=True)
print("\nDuplicates Removed")


```

```python
print("\n Summary")
print(df.describe())

print("\nAverage Total Bill")
print(df['total_bill'].mean())

print("\nMaximum Total Bill")
print(df['total_bill'].max())

print("\nMinimum Total Bill")
print(df['total_bill'].min())

print("\nAverage Tip")
print(df['tip'].mean())

print("\nHighest Tip")
print(df['tip'].max())

dinner = df[df['time'] == 'Dinner']
print("\nDinner Customers")
print(dinner)

lunch = df[df['time'] == 'Lunch']
print("\nLunch Customers")
print(lunch)

print("\nAverage Bill Per Day")
day_analysis = df.groupby('day')['total_bill'].mean()
print(day_analysis)

print("\nAverage Tip by Gender")
gender_tip = df.groupby('sex')['tip'].mean()
print(gender_tip)

print("\nAverage Bill by Smoker")
smoker_bill = df.groupby('smoker')['total_bill'].mean()
print(smoker_bill)

sorted_bill = df.sort_values(
    by='total_bill',
    ascending=False
)

print("\nHighest Bills")
print(sorted_bill)

print("\nTop 5 Highest Bills")
print(sorted_bill.head())

df['tip_percentage'] = (
    df['tip'] / df['total_bill']
) * 100

print("\nTip Percentage")
print(df[['total_bill', 'tip', 'tip_percentage']])

high_tip = df[df['tip_percentage'] > 15]
print("\nCustomers Giving More Than 15% Tip")
print(high_tip)

print("\nFinal Cleaned Dataset")
print(df)
```
